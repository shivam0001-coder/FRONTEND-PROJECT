```python
#1(a)
matrix = []

print("Enter elements for 3x3 matrix:")
for i in range(3):
    row = []
    for j in range(3):
        val = int(input(f"Enter element at position [{i}][{j}]: "))
        row.append(val)
    matrix.append(row)
```

```python
#1(b)
print("\nMatrix is:")
for row in matrix:
    print(row)
```

```python
#2(a)
print("Enter elements for first 2x2 matrix:")
A = []
for i in range(2):
    row = []
    for j in range(2):
        row.append(int(input(f"A[{i}][{j}]: ")))
    A.append(row)
print("\nEnter elements for second 2x2 matrix:")
B = []
for i in range(2):
    row = []
    for j in range(2):
        row.append(int(input(f"B[{i}][{j}]: ")))
    B.append(row)
sum_matrix = []
for i in range(2):
    row = []
    for j in range(2):
        row.append(A[i][j] + B[i][j])
    sum_matrix.append(row)
```

```python
#2(b)
print("\nSum of matrices:")
for row in sum_matrix:
    print(row)
```

```python
#3
A = [
    [1, 2],
    [3, 4],
    [5, 6]
]
transpose = []
for i in range(len(A[0])):
    row = []
    for j in range(len(A)):
        row.append(A[j][i])
    transpose.append(row)

print("Transpose of matrix A:")
for row in transpose:
    print(row)
```

```python
#4
A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [1, 2],
    [3, 4],
    [5, 6]
]

result = [[0 for _ in range(2)] for _ in range(2)]

for i in range(len(A)):
    for j in range(len(B[0])):
        for k in range(len(B)):
            result[i][j] += result[i][j] + A[i][k] * B[k][j]

print("Resultant Matrix:")
for row in result:
    print(row)
```

```python
#5
n = int(input("Enter size of square matrix: "))
matrix = []
print("Enter matrix elements:")
for i in range(n):
    row = []
    for j in range(n):
        row.append(int(input(f"Element [{i}][{j}]: ")))
    matrix.append(row)
is_identity = True
is_diagonal = True
for i in range(n):
    for j in range(n):
        if i == j:
            if matrix[i][j] != 1:
                is_identity = False
        else:
            if matrix[i][j] != 0:
                is_identity = False
                is_diagonal = False
if is_identity:
    print("It is an Identity Matrix")
elif is_diagonal:
    print("It is a Diagonal Matrix")
else:
    print("It is Neither Identity nor Diagonal Matrix")
```
